```python
from app.splitter import pdf_to_chunks
from langchain_community.vectorstores import Chroma
from app.models import embedding_model
from app.models import pipe, prompt_template
class pdf():
    def __init__(self, path):
        self.path = path

file = pdf("bitcoin_es.pdf")

chunks = pdf_to_chunks(file)
vector_db = Chroma.from_documents(documents=chunks, embedding=embedding_model)
```

```python
queries = ["Como funciona bitcoin?", "Que es proof-of-work?", "Como se verifican los pagos?", "Cuales son los incentivos?"]
answers = []
for query in queries:
    results = vector_db.similarity_search(query)
    context = f"{results[0].page_content} {results[1].page_content}"

    prompt = prompt_template.format(context=context, prompt=query)
    respuesta = pipe(prompt)
    answers.append(respuesta)
```

```python
import pandas as pd

df = pd.DataFrame({"query": queries, "answer": answers})
df.to_csv("model_evaluation.csv", index=False)
```
